## Tip Calculator Program

```python
user_name = input("Please enter your name: ")
print(f"Hi {user_name}, welcome to the tip calculator program!")
total_bill_amount = float(input("Please enter your total bill amount: "))
tip_percentage = float(input("Please enter how % do you want to give a tip?: "))
no_of_users_to_split_bill = int(input("Please enter no of person's you want to split your bill amount?: "))
final_bill_amount = (total_bill_amount + (total_bill_amount * (tip_percentage/100))) / no_of_users_to_split_bill
print(f"Your final bill amount for each person's is : {round(final_bill_amount, 2)}")
```

## Improved Program: Tip Calculator - splits a bill among people with a tip.

```python
user_name = input("Please enter your name: ")
print(f"Hi {user_name}, welcome to the tip calculator!")

try:
    bill = float(input("Enter the total bill amount: "))
    tip_percent = float(input("Enter tip percentage (e.g. 10 for 10%): "))
    num_people = int(input("How many people to split the bill? "))

    if num_people <= 0:
        print("Number of people must be at least 1.")
    elif bill < 0 or tip_percent < 0:
        print("Bill and tip cannot be negative.")
    else:
        total_with_tip = bill * (1 + tip_percent / 100)
        per_person = total_with_tip / num_people
        print(f"Total with tip: ${total_with_tip:.2f}")
        print(f"Each person pays: ${per_person:.2f}")

except ValueError:
    print("Invalid input. Please enter numbers only.")
```
