## Find Odd or Even Number Program

```python
print("Welcome to Odd or Even Finder")
number = int(input("Enter a number: "))

if (number % 2) == 0 :
    print(f"The number you have entered {number} is an even number.")
else:
    print(f"The number you have entered {number} is an odd number.")
```

## Improved Program: Odd or Even Finder program

```python
print("Welcome to Odd or Even Finder")

while True:
    user_input = input("Enter a number (or 'q' to quit): ")
    if user_input.lower() == "q":
        print("Goodbye!")
        break
    try:
        number = int(user_input)
        parity = "even" if number % 2 == 0 else "odd"
        print(f"The number you entered ({number}) is {parity}.")
    except ValueError:
        print("Invalid input. Please enter a whole number.")
```

## Fixed & Improved Version: Pizza Order Program

```python
def new_order():
    user_name = input("Enter your name: ")
    print(f"Hi {user_name}, welcome to Python Pizza!")

    try:
        bill = 0
        size = input("What size pizza would you like to order? (S = $15, M = $20, L = $25): ").strip().lower()
        need_pepperoni = input("Would you like to add pepperoni ($2)? (Y or N): ").strip().lower()
        extra_cheese = input("Do you want to add extra cheese ($3)? (Y or N): ").strip().lower()

        # 1. Base price by size
        if size == "s":
            bill = 15
        elif size == "m":
            bill = 20
        elif size == "l":
            bill = 25
        else:
            print("⚠️ Invalid size selected. Please restart.")
            return

        # 2. Independent add-ons
        if need_pepperoni == "y":
            bill += 2

        if extra_cheese == "y":
            bill += 3

        print(f"Your final bill amount is ${bill}.")

    except Exception:
        print("Please enter a valid input.")

new_order()
```

## Pro Level Program: Pizza Order Program

```python
def new_order_pro():
    user_name = input("Enter your name: ")
    print(f"Hi {user_name}, welcome to Python Pizza!")

    menu = {"s": 15, "m": 20, "l": 25, "k": 50}

    try:
        size = input("Size (S/M/L/K): ").strip().lower()
        pepperoni = input("Pepperoni (Y/N): ").strip().lower()
        cheese = input("Extra Cheese (Y/N): ").strip().lower()

        if size not in menu:
            print("Invalid size!")
            return

        bill = menu[size]
        if pepperoni == "y": bill += 2
        if cheese == "y": bill += 3

        print(f"Your total is ${bill}.")
    except Exception:
        print("Invalid input.")

new_order_pro()
```
